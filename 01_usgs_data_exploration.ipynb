# Colorado River Streamflow Analysis

This notebook explores real-time streamflow data from the USGS National Water Information System (NWIS) for the Upper Colorado River Basin. 

We are focusing on our first sentinel gauge: **Colorado River at Lees Ferry, AZ**.

```python
import urllib.request
import json
import pandas as pd

# Lees Ferry USGS Site ID
site_id = '09380000'
# Parameter 00060 is Discharge in cubic feet per second (cfs)
url = f'https://waterservices.usgs.gov/nwis/iv/?format=json&sites={site_id}&parameterCd=00060&siteStatus=all'

req = urllib.request.Request(url, headers={'User-Agent': 'DAT1001-Student-Project/1.0'})

with urllib.request.urlopen(req) as response:
    data = json.loads(response.read().decode('utf-8'))

time_series = data['value']['timeSeries'][0]
site_name = time_series['sourceInfo']['siteName']
latest_flow = time_series['values'][0]['value'][-1]['value']
timestamp = time_series['values'][0]['value'][-1]['dateTime']

print(f"Site: {site_name}")
print(f"Current Flow: {latest_flow} cfs at {timestamp}")
```
